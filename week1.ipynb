```python
import numpy as np
import time
```

```python
size = 1000000

# Python List
python_list_a = list(range(size))
python_list_b = list(range(size))

start_time = time.time()
python_result = [a + b for a, b in zip(python_list_a, python_list_b)]
end_time = time.time()
python_list_time = end_time - start_time
print(f"Time taken for Python List operation: {python_list_time:.6f} seconds")

# NumPy Array
numpy_array_a = np.arange(size)
numpy_array_b = np.arange(size)

start_time = time.time()
numpy_result = numpy_array_a + numpy_array_b
end_time = time.time()
numpy_array_time = end_time - start_time
print(f"Time taken for NumPy Array operation: {numpy_array_time:.6f} seconds")

print(f"\nNumPy array was {python_list_time / numpy_array_time:.2f} times faster than Python list.")
```

```python
import sys

# Assuming 'size' is already defined from the previous cell (1,000,000)
# If not, uncomment the line below:
# size = 1000000

# Python List memory usage
python_list = list(range(size))
python_list_memory = sys.getsizeof(python_list) # Size of the list object itself
# To get a more accurate total memory, we'd also need to sum the size of each element.
# However, for comparison, the object size and knowing elements are separate is sufficient.
# For simple integers, it's often 28 bytes per int on 64-bit systems + list overhead.
python_list_elements_memory = sum(sys.getsizeof(0) for _ in range(size)) # Approximation for int elements

print(f"Python List (object itself) memory: {python_list_memory / (1024*1024):.4f} MB")
print(f"Python List (approx. total with elements) memory: {(python_list_memory + python_list_elements_memory) / (1024*1024):.4f} MB")

# NumPy Array memory usage
numpy_array = np.arange(size)
numpy_array_memory = numpy_array.nbytes # Total bytes consumed by the array data

print(f"NumPy Array memory: {numpy_array_memory / (1024*1024):.4f} MB")

print(f"\nObservation: NumPy arrays are significantly more memory-efficient for numerical data.")
```

```python
import numpy as np

# 1. From a Python List
list_data = [1, 2, 3, 4, 5]
arr_from_list = np.array(list_data)
print(f"From List: {arr_from_list}, Type: {type(arr_from_list)}")

# 2. From a Python Tuple
tuple_data = (6, 7, 8, 9, 10)
arr_from_tuple = np.array(tuple_data)
print(f"From Tuple: {arr_from_tuple}, Type: {type(arr_from_tuple)}")

# 3. From a nested Python List (to create a 2D array)
nested_list_data = [[1, 2, 3], [4, 5, 6]]
arr_2d_from_list = np.array(nested_list_data)
print(f"From Nested List (2D):\n{arr_2d_from_list}, Shape: {arr_2d_from_list.shape}")

# 4. From a Python Dictionary (common approach: convert values or items to list first)
dict_data = {'a': 1, 'b': 2, 'c': 3}
# To create an array of dictionary values:
arr_from_dict_values = np.array(list(dict_data.values()))
print(f"From Dictionary Values: {arr_from_dict_values}")

# To create an array of dictionary items (key-value pairs as tuples, then converted):
arr_from_dict_items = np.array(list(dict_data.items()), dtype=object) # dtype=object might be needed for mixed types
print(f"From Dictionary Items (array of tuples):\n{arr_from_dict_items}")

# 5. From an iterable (e.g., a generator)
gen_data = (x*x for x in range(5))
arr_from_gen = np.array(list(gen_data)) # Convert generator to list first
print(f"From Generator: {arr_from_gen}")
```

```python
import numpy as np

print("### NumPy Random Functions ###")

# np.random.rand() - Uniform random numbers between 0 and 1
# Useful for probabilities or simple random values.
print("\n--- np.random.rand --- ")
print(np.random.rand(3, 2)) # A 3x2 array of random floats

# np.random.randn() - Standard Normal (Gaussian) distribution (mean=0, variance=1)
# Useful for adding noise or sampling from a normal distribution.
print("\n--- np.random.randn --- ")
print(np.random.randn(4)) # 4 random floats from a standard normal distribution

# np.random.randint(low, high, size) - Random integers
# Useful for simulations like dice rolls or picking random indices.
print("\n--- np.random.randint --- ")
print(np.random.randint(1, 10, size=(2, 5))) # A 2x5 array of integers between 1 (inclusive) and 10 (exclusive)

# np.random.choice(array, size, replace=True/False) - Randomly select elements
# Useful for sampling data.
print("\n--- np.random.choice --- ")
my_list = ['A', 'B', 'C', 'D', 'E']
print(np.random.choice(my_list, size=3, replace=False)) # Choose 3 unique elements

# np.random.permutation(array or int) - Randomly reorder sequence
# Useful for shuffling datasets.
print("\n--- np.random.permutation --- ")
print(np.random.permutation(np.arange(5))) # Shuffles numbers 0 to 4
```
