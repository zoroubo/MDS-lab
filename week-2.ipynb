```python
import numpy as np

matrix_2d = np.array([[52, 2, 3],
                      [4, 5, 6]])

scalar_value = 5

result_matrix_multiplication = matrix_2d * scalar_value

print(result_matrix_multiplication)
```

```python
import numpy as np

matrix_a = np.array([[52, 2],
                     [3, 4]])

matrix_b = np.array([[5, 6],
                     [7, 8]])

dot_product_result = np.dot(matrix_a, matrix_b)

print(dot_product_result)
```

```python
import numpy as np

matrix = np.array([[52, 2],
                   [3, 4]])

transposed_matrix = matrix.transpose()

determinant = np.linalg.det(matrix)

print(transposed_matrix)
print(determinant)
```

```python
import csv
import numpy as np

marks_filename = 'student_marks.csv'
num_students = 5
num_subjects = 5

headers = [f'Subject_{i+1}' for i in range(num_subjects)]
data = np.random.randint(40, 100, size=(num_students, num_subjects))

with open(marks_filename, 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(headers)
    writer.writerows(data)

print(f"'{marks_filename}' has been generated.")
```

```python
import numpy as np

marks_dataset = np.genfromtxt('student_marks.csv', delimiter=',', skip_header=1)

print(marks_dataset)
print(marks_dataset.shape)
```

```python
print('Element-wise addition by 7:\n', np.add(marks_dataset, 7))
print('Element-wise square root:\n', np.sqrt(marks_dataset))
print('Element-wise exponential (scaled):\n', np.exp(marks_dataset / 100))
print('Element-wise logarithm (natural):\n', np.log(marks_dataset))
print('Element-wise cosine:\n', np.cos(marks_dataset))
```

```python
print('Overall Mean:', marks_dataset.mean())
print('Mean per Subject (columns):', marks_dataset.mean(axis=0))
print('Mean per Student (rows):', marks_dataset.mean(axis=1))

print('Overall Sum:', marks_dataset.sum())
print('Sum per Subject (columns):', marks_dataset.sum(axis=0))

print('Overall Minimum:', marks_dataset.min())
print('Overall Maximum:', marks_dataset.max())

print('Overall Standard Deviation:', marks_dataset.std())
print('Overall Variance:', marks_dataset.var())
```

```python
import numpy as np

# Assuming marks_dataset is already loaded from 'student_marks.csv'
# marks_dataset = np.genfromtxt('student_marks.csv', delimiter=',', skip_header=1)

print('*** Statistical Operations ***')

print('Median of all elements:', np.median(marks_dataset))
print('Median per Subject (columns):', np.median(marks_dataset, axis=0))
print('Median per Student (rows):', np.median(marks_dataset, axis=1))

print('Standard Deviation per Subject (columns):', np.std(marks_dataset, axis=0))
print('Standard Deviation per Student (rows):', np.std(marks_dataset, axis=1))

print('Variance per Subject (columns):', np.var(marks_dataset, axis=0))
print('Variance per Student (rows):', np.var(marks_dataset, axis=1))

print('Minimum per Subject (columns):', np.min(marks_dataset, axis=0))
print('Maximum per Subject (columns):', np.max(marks_dataset, axis=0))

print('Percentile 25 (Q1) of all elements:', np.percentile(marks_dataset, 25))
print('Percentile 75 (Q3) of all elements:', np.percentile(marks_dataset, 75))
```

```python
import numpy as np

print('*** Comparison Operations ***')

print('Marks greater than 70:\n', marks_dataset > 70)
print('Number of marks greater than 70:', np.sum(marks_dataset > 70))

print('Marks less than or equal to 50:\n', marks_dataset <= 50)

comparison_array = np.full_like(marks_dataset, 75)
print('Marks equal to 75 (element-wise comparison with a full array):\n', marks_dataset == comparison_array)

print('Conditional selection (marks above 70, otherwise 0):\n', np.where(marks_dataset > 70, marks_dataset, 0))

print('Check if all marks in a row are greater than 40:\n', np.all(marks_dataset > 40, axis=1))
print('Check if any mark in a column is less than 50:\n', np.any(marks_dataset < 50, axis=0))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import urllib.request
from PIL import Image
from skimage import data

# --- Step 1: Load the Image ---
# Using a built-in animal image from scikit-image for reliability
# img = data.chelsea()

# To load your own image, first upload it to the Colab environment (e.g., to /content/).
# Then, replace the path below with the correct path to your uploaded image.
# For example, if you upload 'catimg.jpg' directly, it will be in '/content/catimg.jpg'.
image_path = '/content/catimg.jpg'

# Load the image using PIL and convert to a NumPy array
img = np.array(Image.open(image_path))

# Display the original image
plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.title('My Own Image (loaded locally in Colab)')
plt.axis('off')
plt.show()
```

```python
# These coordinates are illustrative and will need adjustment for your specific image.
# Example coordinates (you will need to change these based on your image):
row_start = 200
row_end = 400
col_start = 100
col_end = 500


eyes_cropped = img[row_start:row_end, col_start:col_end]

plt.figure(figsize=(6, 6))
plt.imshow(eyes_cropped)
plt.title('Cropped to Eyes (Adjust Coordinates!)')
plt.axis('off')
plt.show()
```

### Step 3: Flip operations

```python
# Flip horizontally (mirror image)
eyes_flipped_horizontal = eyes_cropped[:, ::-1]
plt.figure(figsize=(6, 6))
plt.imshow(eyes_flipped_horizontal)
plt.title('Eyes Flipped Horizontally')
plt.axis('off')
plt.show()

# Flip vertically (upside down)
eyes_flipped_vertical = eyes_cropped[::-1, :]
plt.figure(figsize=(6, 6))
plt.imshow(eyes_flipped_vertical)
plt.title('Eyes Flipped Vertically')
plt.axis('off')
plt.show()
```
